```python

# REAL-WORLD DATA SCIENCE PROJECT
# RETAIL SALES DATA ANALYSIS USING PYTHON

# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
np.random.seed(42)

# 2. CREATE RETAIL SALES DATASET
n = 1000

dates = pd.date_range(
    start="2025-01-01",
    end="2025-12-31",
    periods=n
)

products = ['Laptop', 'Mobile', 'Headphones',
            'Keyboard', 'Mouse', 'Monitor']

categories = {
    'Laptop': 'Electronics',
    'Mobile': 'Electronics',
    'Headphones': 'Accessories',
    'Keyboard': 'Accessories',
    'Mouse': 'Accessories',
    'Monitor': 'Electronics'
}

prices = {
    'Laptop': 50000,
    'Mobile': 20000,
    'Headphones': 1500,
    'Keyboard': 1000,
    'Mouse': 500,
    'Monitor': 12000
}

df = pd.DataFrame({
    'Order_ID': range(1, n + 1),
    'Order_Date': dates,
    'Customer_ID': np.random.randint(1001, 1201, n),
    'Product': np.random.choice(products, n),
    'Quantity': np.random.randint(1, 6, n),
    'Payment_Method': np.random.choice(
        ['UPI', 'Credit Card', 'Debit Card', 'Cash'], n
    ),
    'City': np.random.choice(
        ['Hyderabad', 'Mumbai', 'Delhi', 'Chennai', 'Bangalore'], n
    )
})

# Add category and price
df['Category'] = df['Product'].map(categories)
df['Unit_Price'] = df['Product'].map(prices)

# Calculate total sales
df['Total_Sales'] = df['Quantity'] * df['Unit_Price']

# Introduce a few data quality issues
df.loc[5, 'Quantity'] = np.nan
df.loc[10, 'City'] = np.nan
df = pd.concat([df, df.iloc[[0, 1]]], ignore_index=True)

# 3. DISPLAY DATASET
print("FIRST FIVE RECORDS")
display(df.head())

print("\nDATASET SHAPE:", df.shape)

# 4. DATA INFORMATION
print("\nDATASET INFORMATION")
df.info()

# 5. STATISTICAL SUMMARY
print("\nSTATISTICAL SUMMARY")
display(df.describe())

# 6. CHECK MISSING VALUES
print("\nMISSING VALUES")
print(df.isnull().sum())

# 7. REMOVE DUPLICATES
print("\nDUPLICATES:", df.duplicated().sum())
df.drop_duplicates(inplace=True)

# 8. HANDLE MISSING VALUES
df['Quantity'] = df['Quantity'].fillna(
    df['Quantity'].median()
)

df['City'] = df['City'].fillna(
    df['City'].mode()[0]
)

# 9. RECOMPUTE SALES
df['Total_Sales'] = df['Quantity'] * df['Unit_Price']

# 10. CREATE MONTH COLUMN
df['Month'] = df['Order_Date'].dt.month_name()
df['Month_Number'] = df['Order_Date'].dt.month

# 11. TOTAL REVENUE
total_revenue = df['Total_Sales'].sum()

print("\nTOTAL REVENUE:", total_revenue)

# 12. TOTAL ORDERS
print("TOTAL ORDERS:", df['Order_ID'].nunique())

# 13. TOP SELLING PRODUCTS
product_sales = df.groupby(
    'Product'
)['Total_Sales'].sum().sort_values(ascending=False)

print("\nPRODUCT-WISE SALES")
print(product_sales)

plt.figure(figsize=(9, 5))
product_sales.plot(kind='bar')
plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=30)
plt.show()

# 14. MONTHLY SALES ANALYSIS
monthly_sales = df.groupby(
    'Month_Number'
)['Total_Sales'].sum()

month_names = [
    'Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
    'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'
]

monthly_sales.index = [
    month_names[i - 1] for i in monthly_sales.index
]

plt.figure(figsize=(10, 5))
monthly_sales.plot(kind='line', marker='o')
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()

# 15. CATEGORY-WISE SALES
category_sales = df.groupby(
    'Category'
)['Total_Sales'].sum()

plt.figure(figsize=(7, 5))
category_sales.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Category-wise Sales")
plt.ylabel("")
plt.show()

# 16. CITY-WISE REVENUE
city_sales = df.groupby(
    'City'
)['Total_Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(9, 5))
sns.barplot(
    x=city_sales.index,
    y=city_sales.values
)
plt.title("City-wise Revenue")
plt.xlabel("City")
plt.ylabel("Revenue")
plt.xticks(rotation=20)
plt.show()

# 17. PAYMENT METHOD ANALYSIS
payment_counts = df['Payment_Method'].value_counts()

plt.figure(figsize=(8, 5))
payment_counts.plot(
    kind='bar',
    color='skyblue'
)
plt.title("Payment Method Distribution")
plt.xlabel("Payment Method")
plt.ylabel("Number of Orders")
plt.show()

# 18. CUSTOMER SPENDING ANALYSIS
customer_sales = df.groupby(
    'Customer_ID'
)['Total_Sales'].sum().sort_values(ascending=False)

print("\nTOP 10 CUSTOMERS")
print(customer_sales.head(10))

# 19. DAILY SALES DISTRIBUTION
plt.figure(figsize=(9, 5))
sns.histplot(
    df['Total_Sales'],
    bins=30,
    kde=True
)
plt.title("Sales Distribution")
plt.xlabel("Sales Amount")
plt.show()

# 20. CORRELATION HEATMAP
plt.figure(figsize=(7, 5))
sns.heatmap(
    df[['Quantity', 'Unit_Price', 'Total_Sales']].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)
plt.title("Sales Correlation Heatmap")
plt.show()

# 21. KEY BUSINESS INSIGHTS
print("\n========== BUSINESS INSIGHTS ==========")

print("Total Revenue:", round(total_revenue, 2))

print("Average Order Value:",
      round(df['Total_Sales'].mean(), 2))

print("Best Selling Product:",
      product_sales.idxmax())

print("Highest Revenue City:",
      city_sales.idxmax())

print("Most Used Payment Method:",
      payment_counts.idxmax())

print("Best Revenue Category:",
      category_sales.idxmax())

# 22. SAVE CLEANED DATASET
df.to_csv("Retail_Sales_Cleaned.csv", index=False)

# 23. GENERATE PROJECT REPORT
report = f"""
RETAIL SALES DATA ANALYSIS REPORT

PROJECT OVERVIEW:
Analysis of retail sales transactions using Python.

DATASET:
Total Records: {len(df)}
Unique Orders: {df['Order_ID'].nunique()}
Unique Customers: {df['Customer_ID'].nunique()}

BUSINESS RESULTS:
Total Revenue: {total_revenue:.2f}
Average Order Value: {df['Total_Sales'].mean():.2f}
Best Selling Product: {product_sales.idxmax()}
Highest Revenue City: {city_sales.idxmax()}
Most Used Payment Method: {payment_counts.idxmax()}

PRODUCT SALES:
{product_sales.to_string()}

CITY SALES:
{city_sales.to_string()}

CATEGORY SALES:
{category_sales.to_string()}

CONCLUSION:
The retail dataset was cleaned and analyzed using Python.
Monthly sales, product performance, customer spending,
city revenue and payment methods were examined.

These insights can help businesses understand sales
patterns and make better business decisions.

NOTE:
This project uses a synthetic sample dataset for educational
purposes. It does not represent actual company transactions.
"""

with open("Retail_Sales_Report.txt", "w") as file:
    file.write(report)

# 24. DOWNLOAD OUTPUT FILES
from google.colab import files

files.download("Retail_Sales_Cleaned.csv")
files.download("Retail_Sales_Report.txt")

print("\nRETAIL SALES ANALYSIS COMPLETED SUCCESSFULLY!")
```
