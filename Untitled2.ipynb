```python

# EXPLORATORY DATA ANALYSIS (EDA) PROJECT
# Wine Dataset Analysis

# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_wine

sns.set_theme(style="whitegrid")

# 2. LOAD DATASET
wine = load_wine()

df = pd.DataFrame(
    wine.data,
    columns=wine.feature_names
)

df['target'] = wine.target
df['wine_class'] = df['target'].map(
    {i: name for i, name in enumerate(wine.target_names)}
)

print("FIRST FIVE ROWS")
display(df.head())

print("\nDATASET SHAPE:", df.shape)
print("\nCOLUMN NAMES:")
print(df.columns.tolist())

# 3. DATASET INFORMATION
print("\nDATASET INFORMATION")
df.info()

# 4. STATISTICAL SUMMARY
print("\nSTATISTICAL SUMMARY")
display(df.describe())

# 5. CHECK MISSING VALUES
print("\nMISSING VALUES")
print(df.isnull().sum())

# 6. CHECK DUPLICATES
print("\nDUPLICATE ROWS:", df.duplicated().sum())

# 7. REMOVE DUPLICATES
df = df.drop_duplicates()

# 8. CLASS DISTRIBUTION
print("\nWINE CLASS DISTRIBUTION")
print(df['wine_class'].value_counts())

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x='wine_class')
plt.title("Wine Class Distribution")
plt.xlabel("Wine Class")
plt.ylabel("Count")
plt.show()

# 9. HISTOGRAMS
df[wine.feature_names].hist(
    figsize=(16, 12),
    bins=20,
    edgecolor='black'
)

plt.suptitle("Distribution of Wine Features")
plt.tight_layout()
plt.show()

# 10. BOXPLOTS
plt.figure(figsize=(12, 6))
sns.boxplot(data=df[wine.feature_names])
plt.title("Boxplot of Wine Features")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

# 11. CORRELATION ANALYSIS
correlation = df[wine.feature_names].corr()

plt.figure(figsize=(13, 9))
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap")
plt.show()

# 12. TOP CORRELATED FEATURES
corr_pairs = correlation.abs().unstack()

corr_pairs = corr_pairs[
    corr_pairs.index.get_level_values(0) !=
    corr_pairs.index.get_level_values(1)
]

corr_pairs = corr_pairs.drop_duplicates().sort_values(
    ascending=False
)

print("\nTOP 10 CORRELATED FEATURE PAIRS")
print(corr_pairs.head(10))

# 13. FEATURE COMPARISON
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=df,
    x='alcohol',
    y='color_intensity',
    hue='wine_class',
    s=70
)

plt.title("Alcohol vs Color Intensity")
plt.show()

# 14. CLASS-WISE FEATURE ANALYSIS
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x='wine_class',
    y='alcohol'
)

plt.title("Alcohol Content by Wine Class")
plt.show()

# 15. PAIR PLOT
sns.pairplot(
    df,
    vars=[
        'alcohol',
        'malic_acid',
        'color_intensity',
        'proline'
    ],
    hue='wine_class'
)

plt.show()

# 16. KEY INSIGHTS
print("\n========== KEY INSIGHTS ==========")

print("Total Records:", len(df))
print("Total Features:", len(wine.feature_names))
print("Number of Wine Classes:", df['wine_class'].nunique())
print("Missing Values:", df.isnull().sum().sum())

print("\nAverage Feature Values:")
display(df[wine.feature_names].mean().sort_values(
    ascending=False
).head(5))

print("\nMost Strongly Correlated Features:")
print(corr_pairs.head(5))

# 17. SAVE ANALYSIS DATA
df.to_csv("EDA_Wine_Analysis.csv", index=False)

# 18. GENERATE EDA REPORT
report = f"""
EXPLORATORY DATA ANALYSIS REPORT

Project: Wine Dataset Analysis

1. Dataset Overview
Total Records: {len(df)}
Total Features: {len(wine.feature_names)}
Wine Classes: {df['wine_class'].nunique()}

2. Data Quality
Missing Values: {df.isnull().sum().sum()}
Duplicate Records Remaining: {df.duplicated().sum()}

3. Statistical Analysis
{df[wine.feature_names].describe().to_string()}

4. Wine Class Distribution
{df['wine_class'].value_counts().to_string()}

5. Top Correlated Features
{corr_pairs.head(10).to_string()}

6. Conclusion
The dataset was explored using statistical summaries
and visualizations. Feature distributions, correlations,
and differences between wine classes were analyzed.

The analysis helps identify patterns and relationships
among the chemical properties of wine.
"""

with open("EDA_Summary_Report.txt", "w") as file:
    file.write(report)

print("\nEDA REPORT GENERATED SUCCESSFULLY!")
print("CSV: EDA_Wine_Analysis.csv")
print("Report: EDA_Summary_Report.txt")

# 19. DOWNLOAD FILES
from google.colab import files

files.download("EDA_Wine_Analysis.csv")
files.download("EDA_Summary_Report.txt")

print("\nPROJECT COMPLETED SUCCESSFULLY!")
```
